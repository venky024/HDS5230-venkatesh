```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Define a list of model names used in the evaluation
models = ['Dummy', 'KNN-1', 'KNN-2', 'KNN-3-1', 'KNN-3-2']
stages = ['Dummy', 'KNN Training', 'KNN Prediction', 'KNN Retraining', 'KNN Retrained Prediction']
accuracies = [0.5075, 0.895, 0.963, 0.721, 0.965]
recalls = [0.5327, 0.9045, 0.9061, 1.0, 0.8873]
precisions = [0.5048, 0.8867, 0.9407, 0.4163, 0.913]
```

```python
# Create an array of evenly spaced values representing the index positions of each model
x = np.arange(len(models))
```

```python
plt.figure(figsize=(12, 8))

plt.plot(x, accuracies, 'o-', linewidth=2, label='Accuracy')
plt.plot(x, recalls, 'o-', linewidth=2, label='Recall')
plt.plot(x, precisions, 'o-', linewidth=2, label='Precision')
plt.xticks(x, stages, rotation=15)
plt.grid(True, linestyle='--', alpha=0.7)
plt.xlabel('Model Training Stage')
plt.ylabel('Score')
plt.title('Model Performance Metrics Across Different Training Stages')
plt.legend()

for i, v in enumerate(accuracies):
    plt.text(i, v+0.02, f'{v:.3f}', ha='center', va='bottom', color='blue')
for i, v in enumerate(recalls):
    plt.text(i, v+0.02, f'{v:.3f}', ha='center', va='bottom', color='green')
for i, v in enumerate(precisions):
    plt.text(i, v-0.03, f'{v:.3f}', ha='center', va='top', color='red')
plt.axhline(y=0.5, color='gray', linestyle='--', alpha=0.5, label='Baseline')
plt.ylim(0.4, 1.1)
plt.tight_layout()
plt.show()
```
